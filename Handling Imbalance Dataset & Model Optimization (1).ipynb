# Import Libraries

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,MinMaxScaler,RobustScaler
```

# Import Dataset

```python
from sklearn import datasets
Breast_cancer_dataset=datasets.load_breast_cancer()
Breast_cancer_dataset
```

```python
breast_cancer_data_df=pd.DataFrame(Breast_cancer_dataset.data,columns=Breast_cancer_dataset.feature_names)
```

```python
breast_cancer_data_df.head(10)
```

```python
print(Breast_cancer_dataset.DESCR)
```

```python
breast_cancer_data_df["Target"]=Breast_cancer_dataset["target"]
```

```python
breast_cancer_data_df
```

# Understanding Datasets


```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

```python
print(breast_cancer_data_df.columns.tolist())
```

```python
X=breast_cancer_data_df.drop(labels="Target",axis=1)
X
```

```python
y=breast_cancer_data_df["Target"]
y
```

 # Model Building

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report,confusion_matrix
from matplotlib import pyplot as plt
```

```python
y.value_counts()
```

```python
breast_cancer_data_df["Target"].value_counts().plot.pie(
    autopct="%1.1f%%",
    labels=["Benign", "Malignant"],
    title="Whole Data Distribution"
)
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split=train_test_split(X,y,test_size=0.20,random_state=1234,
    shuffle=True,
    stratify=y,
)
```

```python
print("Training Sample Shape:",X_train.shape,y_train.shape)
print("Test Sample Shape:",X_test.shape,y_test.shape)
```

```python
y_train.value_counts().sort_index().plot.pie(
    autopct="%1.1f%%",
    labels=["Malignant", "Benign"],
    title="Training Data Distribution"
)
plt.ylabel("count")
plt.show()
```

```python
y_test.value_counts().sort_index().plot.pie(
    autopct="%1.1f%%",
    labels=["Malignant", "Benign"],
    title="Testing Data Distribution"
)
plt.ylabel("count")
plt.show()
```

```python
%%time
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier(criterion="gini",max_depth=3,class_weight={0:2,1:1})
dt_model.fit(X_train,y_train)
```

# Model Training

```python
y_pred_train=dt_model.predict(X_train)
```

```python
y_pred_test=dt_model.predict(X_test)
```

# Model Evalution

```python
print(confusion_matrix(y_train,y_pred_train))
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```

```python
print(classification_report(y_test,y_pred_test))
```
